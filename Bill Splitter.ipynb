## Bill Splitter
In this workshop, we will use numbers and mathematical operations to build it.
This tool calculates how much each person should owe/pay after adding meal costs and tip.

* To keep track of the total cost, we create a variable named **'running_total'** and assign it the value 0.

```python
running_total = 0
```

* Let us suppose that there are a total of four friends, so we declare another variable as `num_of_friends`:

```python
num_of_friends = 4
```

* Suppose there are 4 courses of meals, namely appetizers, main courses, desserts, and drinks. Now we are going to declare variables for these courses.
* We assign them values/amounts as decimal numbers (float values) because we consider that these amounts/values involve cents for the sake of our practice.

```python
appetizers = 37.89
main_courses = 57.34
desserts = 39.39
drinks = 64.21
```

* After storing the cost of each individual meal course, we are going to find the total bill so far by adding the values of all courses to the variable "running_total" by using the `+=` augmented operator.
* We then print our result as 'Total bill so far:'

```python
running_total += appetizers + main_courses + desserts + drinks
print("Total Bill so far:", running_total)
```

* Again, suppose that the friends like the service, so they decided to leave a tip amounting to 25% of the total bill so far.
* So we are going to include the tip by multiplying the decimal equivalent of the tip percentage by `running_total`.
* Then we print our result as **'Tip amount'**.

```python
tip = running_total * 0.25
print("Tip amount:" , tip)
```

* After calculating the tip amount, we can now calculate the total bill with tip.
* This can be done by adding the `tip` variable to `running_total` by again using the augmented operator.
* Then we print our result as **'Total with tip'**.

```python
running_total += tip
print("Total with tip:", running_total)
```

* With the tip now included, we have the final amount for the entire group of friends.
* We need to determine how much each person should owe to pay by dividing the variable `running_total` (tip already included) by `num_of_friends`.
* Then we print our result as **'Bill per person'**.

```python
final_bill = running_total / num_of_friends
print('Bill per person:', final_bill)
```

* Finally, we have found the amount that one person in the group should owe to pay.
* The acquired result includes many decimal places, so we want to round it up to two decimal places using the `round()` function.
* We then print our result as **'Each person Pays'**.

```python
each_pays = round(final_bill, 2)
print('Each person pays:', each_pays)
```

## Full Code for the Bill Splitter Workshop

```python
running_total = 0
num_of_friends = 4
appetizers = 37.89
main_courses = 57.34
desserts = 39.39
drinks = 64.21
running_total += appetizers + main_courses + desserts + drinks
print("Total Bill so far:", running_total)
tip = running_total * 0.25
print('Tip amount:', tip)
running_total += tip
print("Total with tip:", running_total)
final_bill = running_total / num_of_friends
print('Bill per person:', final_bill)
each_pays = round(final_bill, 2)
print('Each person pays:', each_pays)
```
